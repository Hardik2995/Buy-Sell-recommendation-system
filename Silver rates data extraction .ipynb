```python
data = browser.page_source
soup = bs(data, 'html.parser')
temp = soup.find_all("table", attrs={"class": "datatable_table__D_jso datatable_table--border__B_zW0 datatable_table--mobile-basic__W2ilt datatable_table--freeze-column__7YoIE"})
len(temp)
```

```python
date = []
price = []
```

```python
date_temp = soup.find_all("td", attrs={"class": "datatable_cell__0y0eu font-bold"})
price_down_temp = soup.find_all("td", attrs={"class": "datatable_cell__0y0eu datatable_cell--align-end__fwomz datatable_cell--down__d61fR"})
price_up_temp = soup.find_all("td", attrs={"class": 'datatable_cell__0y0eu datatable_cell--align-end__fwomz datatable_cell--up__Gt_ed'})
```

```python
for i in range(0,1318):
    temp=date_temp[i].text.strip()
    print (temp)
    date.append(temp)
```
